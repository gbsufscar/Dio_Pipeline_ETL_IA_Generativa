### Relatório de Pedidos de Exportação - Pipeline ETL
Este notebook demonstra um pipeline ETL para acumular as toneladas de exportação de cada país.

## 1. Extract (Extração)
Leitura dos dados de pedidos de exportação de um arquivo CSV

```python
# Bibliotecas necessárias
import pandas as pd
import json
from pathlib import Path
```

```python
# Localização dos dados (ajuste se necessário)
arquivo_pedidos = "pedidos_exportacao_sample.csv"
print(f"Arquivo de entrada: {arquivo_pedidos}")
```

```python
# Lê o arquivo CSV com dados de pedidos de exportação
df_pedidos = pd.read_csv(arquivo_pedidos)
print(f"Total de pedidos carregados: {len(df_pedidos)}")
df_pedidos.head()
```

## 2. Transform (Transformação)
Acumular as toneladas de exportação de cada país em um dicionário

```python
# Acumule as toneladas de exportação de cada país no dicionário
toneladas_por_pais = {}

for index, pedido in df_pedidos.iterrows():
    pais = pedido['pais']
    toneladas = pedido['toneladas']
    
    # Se o país já existe no dicionário, acumula as toneladas
    if pais in toneladas_por_pais:
        toneladas_por_pais[pais] += toneladas
    else:
        # Se o país não existe, cria a entrada
        toneladas_por_pais[pais] = toneladas

print("\nToneladas acumuladas por país:")
for pais, total in toneladas_por_pais.items():
    print(f"{pais}: {total:.2f} toneladas")
```

```python
# Alternativa: Usando groupby do Pandas (método mais eficiente)
toneladas_por_pais_pandas = df_pedidos.groupby('pais')['toneladas'].sum().to_dict()

print("\nVerificação usando Pandas groupby:")
for pais, total in toneladas_por_pais_pandas.items():
    print(f"{pais}: {total:.2f} toneladas")
```

```python
# Criar DataFrame com o resultado para análise
df_resultado = pd.DataFrame([
    {'pais': pais, 'total_toneladas': total}
    for pais, total in toneladas_por_pais.items()
]).sort_values('total_toneladas', ascending=False)

print("\nResumo por país (ordenado por toneladas):")
df_resultado
```

## 3. Load (Carregamento)
Salvar o resultado em arquivo JSON e CSV

```python
# Salvar dicionário em formato JSON
arquivo_json = "toneladas_por_pais.json"
with open(arquivo_json, 'w', encoding='utf-8') as f:
    json.dump(toneladas_por_pais, f, ensure_ascii=False, indent=2)

print(f"Resultado salvo em: {arquivo_json}")
```

```python
# Salvar DataFrame em formato CSV
arquivo_csv = "relatorio_exportacao_por_pais.csv"
df_resultado.to_csv(arquivo_csv, index=False, encoding='utf-8')

print(f"Relatório salvo em: {arquivo_csv}")
```

## 4. Análise Adicional
Estatísticas e visualização dos dados

```python
# Estatísticas gerais
print("=" * 50)
print("ESTATÍSTICAS DE EXPORTAÇÃO")
print("=" * 50)
print(f"\nTotal de pedidos processados: {len(df_pedidos)}")
print(f"Número de países: {len(toneladas_por_pais)}")
print(f"Total de toneladas exportadas: {sum(toneladas_por_pais.values()):.2f}")
print(f"\nPaís com maior volume de exportação: {df_resultado.iloc[0]['pais']} ({df_resultado.iloc[0]['total_toneladas']:.2f} toneladas)")
print(f"Média de toneladas por país: {df_resultado['total_toneladas'].mean():.2f}")
print("=" * 50)
```

## Conclusão

Este notebook demonstra um pipeline ETL completo para:
- **Extrair** dados de pedidos de exportação de um arquivo CSV
- **Transformar** os dados acumulando as toneladas por país em um dicionário
- **Carregar** os resultados em formatos JSON e CSV

O dicionário `toneladas_por_pais` contém o total acumulado de toneladas exportadas para cada país, resolvendo o TODO proposto.
